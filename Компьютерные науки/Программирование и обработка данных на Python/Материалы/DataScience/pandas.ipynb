## 2. Series — одномерная структура

**Series** = массив с подписями (индексом). Как столбец в таблице.

Документация: [pandas.Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.html)

````python
# Создание из списка
s1 = pd.Series([10, 20, 30, 40])
print('Из списка:\n', s1)
print(f'\nindex: {s1.index}')
print(f'values: {s1.values}')  # numpy-массив под капотом
print(f'dtype: {s1.dtype}')
````

````python
# Создание с именованным индексом
s2 = pd.Series(
    [95, 87, 73, 64],
    index=['Алиса', 'Боб', 'Чарли', 'Дейв'],
    name='Баллы'
)
print(s2)
print(f'\ns2["Алиса"] = {s2["Алиса"]}')
print(f's2[0] = {s2.iloc[0]}')
````

````python
# Создание из словаря
population = pd.Series({
    'Москва': 12_655_050,
    'Санкт-Петербург': 5_384_342,
    'Новосибирск': 1_625_631,
    'Екатеринбург': 1_539_371,
}, name='Население')
print(population)
````

````python
# Операции с Series — векторизованные, как в NumPy
s = pd.Series([10, 20, 30, 40, 50])

print('s * 2:\n', s * 2)
print('\ns > 25:\n', s > 25)
print('\ns[s > 25]:\n', s[s > 25])
print(f'\nsum={s.sum()}, mean={s.mean():.1f}, std={s.std():.2f}')
````

## 6. Выбор данных: loc, iloc, at, iat

Главное различие:
- **`loc`** — по **меткам** (именам)
- **`iloc`** — по **позициям** (номерам)

Документация: [Indexing and selecting data](https://pandas.pydata.org/docs/user_guide/indexing.html)

````python
# Подготовка
df_demo = pd.DataFrame({
    'name': ['Алиса', 'Боб', 'Чарли', 'Дейв', 'Ева'],
    'age': [22, 25, 23, 27, 21],
    'score': [95.5, 87.0, 73.2, 64.8, 91.3],
}, index=['a', 'b', 'c', 'd', 'e'])  # строковый индекс
print(df_demo)
print()
````

````python
# === Выбор столбцов ===
print('df["name"]:', type(df_demo['name']))           # Series
print(df_demo['name'], '\n')

print('df[["name", "score"]]:', type(df_demo[['name', 'score']]))  # DataFrame
print(df_demo[['name', 'score']])
````

````python
# === loc — по МЕТКАМ ===
print('Одна строка (Series):')
print(df_demo.loc['a'], '\n')             # строка 'a' как Series

print('Элемент:', df_demo.loc['a', 'score'])     # 95.5

print('\nСрез строк (ВКЛЮЧАЯ обе границы!):')
print(df_demo.loc['b':'d'])                       # строки b, c, d

print('\nНесколько строк + столбцов:')
print(df_demo.loc[['a', 'c', 'e'], ['name', 'score']])
````

````python
# === iloc — по ПОЗИЦИЯМ (как в NumPy) ===
print('Строка 0:')
print(df_demo.iloc[0], '\n')             # первая строка

print('Элемент [0, 2]:', df_demo.iloc[0, 2])  # 95.5

print('\nСрез [1:3] (НЕ включая правую границу):')
print(df_demo.iloc[1:3])                  # строки 1, 2

print('\nПоследние 2 строки, первые 2 столбца:')
print(df_demo.iloc[-2:, :2])
````

````python
# === at, iat — быстрый доступ к ОДНОМУ элементу ===
print('at:', df_demo.at['a', 'name'])     # по метке
print('iat:', df_demo.iat[0, 0])          # по позиции
# Быстрее чем loc/iloc для одиночных элементов
````

<a id='7'></a>
## 7. Фильтрация и query()

Документация: [Boolean indexing](https://pandas.pydata.org/docs/user_guide/indexing.html#boolean-indexing)

````python
df = pd.DataFrame({
    'name': ['Алиса', 'Боб', 'Чарли', 'Дейв', 'Ева', 'Фрэнк'],
    'age': [22, 25, 23, 27, 21, 26],
    'score': [95.5, 87.0, 73.2, 64.8, 91.3, 55.0],
    'city': ['Москва', 'СПб', 'Москва', 'СПб', 'Казань', 'Москва'],
})

# Простая фильтрация
print('Score > 80:\n', df[df['score'] > 80], '\n')

# Комбинация условий: & (и), | (или), ~ (не)
mask = (df['score'] > 70) & (df['city'] == 'Москва')
print('Score > 70 AND Москва:\n', df[mask], '\n')

# isin — принадлежность списку
print('Города Москва или Казань:\n', df[df['city'].isin(['Москва', 'Казань'])], '\n')

# between
print('Age 22-25:\n', df[df['age'].between(22, 25)])
````

````python
# === query() — SQL-подобный синтаксис ===
# Короче и читабельнее для сложных условий

print(df.query('score > 80'), '\n')
print(df.query('score > 70 and city == "Москва"'), '\n')
print(df.query('city in ["Москва", "Казань"]'), '\n')

# Можно использовать переменные через @
threshold = 80
print(df.query('score > @threshold'))
````

<a id='8'></a>
## 8. Добавление и удаление столбцов/строк

````python
df = pd.DataFrame({
    'name': ['Алиса', 'Боб', 'Чарли'],
    'score': [95, 87, 73],
})

# Добавить столбец
df['passed'] = df['score'] >= 80
df['grade'] = pd.cut(df['score'], bins=[0, 60, 75, 90, 100], labels=['F', 'C', 'B', 'A'])
print('Добавлены столбцы:\n', df, '\n')

# assign — возвращает НОВЫЙ DataFrame (удобно для цепочек)
df2 = df.assign(
    score_norm=lambda x: x['score'] / 100,
    bonus=10
)
print('assign:\n', df2, '\n')

# insert — вставка в конкретную позицию
df_copy = df.copy()
df_copy.insert(1, 'id', [1, 2, 3])  # столбец 'id' на позицию 1
print('insert:\n', df_copy)
````

````python
# Удаление
df = pd.DataFrame({'A': [1, 2], 'B': [3, 4], 'C': [5, 6]})

# Удалить столбец
df2 = df.drop(columns=['C'])              # новый DataFrame
print('drop столбца:\n', df2, '\n')

# Удалить строку
df3 = df.drop(index=[0])                   # удалить строку 0
print('drop строки:\n', df3, '\n')

# del и pop
df_copy = df.copy()
del df_copy['B']                           # in-place
print('del:\n', df_copy)
````

## 9. Типы данных и преобразование

Правильные типы = меньше памяти + правильные операции.

Документация: [dtypes](https://pandas.pydata.org/docs/user_guide/basics.html#dtypes)

````python
df = pd.DataFrame({
    'id': ['1', '2', '3'],           # строки, должны быть числа
    'price': ['10.5', '20.3', 'N/A'],  # строки с мусором
    'date': ['2024-01-15', '2024-02-20', '2024-03-10'],  # строки, должны быть даты
    'flag': [1, 0, 1],               # числа, лучше bool
})
print('До:\n', df.dtypes, '\n')

# astype — жёсткое преобразование
df['id'] = df['id'].astype(int)
df['flag'] = df['flag'].astype(bool)

# to_numeric — с обработкой ошибок
df['price'] = pd.to_numeric(df['price'], errors='coerce')  # 'N/A' -> NaN

# to_datetime
df['date'] = pd.to_datetime(df['date'])

print('После:\n', df.dtypes)
print('\n', df)
````

````python
# Nullable типы (поддерживают NaN для целых чисел!)
s = pd.array([1, 2, None, 4], dtype=pd.Int64Dtype())  # или 'Int64'
print(f'Nullable Int64: {s}')     # [1, 2, <NA>, 4] — не float!

# Строковый тип 'string' вместо 'object'
s_str = pd.Series(['hello', 'world', None], dtype='string')
print(f'String type: {s_str.dtype}')  # string, не object
````

<a id='10'></a>
## 10. Пропущенные значения (NaN)

Реальные данные ВСЕГДА содержат пропуски. Нужно уметь с ними работать.

Документация: [Working with missing data](https://pandas.pydata.org/docs/user_guide/missing_data.html)

````python
df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, np.nan, 5],
    'C': [1, 2, 3, 4, 5],
    'D': ['x', np.nan, 'z', np.nan, 'w'],
})
print(df, '\n')

# Обнаружение
print('isna():\n', df.isna(), '\n')          # True где пропуск
print('Пропусков в каждом столбце:\n', df.isna().sum(), '\n')
print(f'Всего пропусков: {df.isna().sum().sum()}')
print(f'% пропусков:\n{(df.isna().mean() * 100).round(1)}')
````

````python
# === Удаление пропусков ===
print('dropna() — удалить строки с ЛЮБЫМ NaN:')
print(df.dropna(), '\n')

print('dropna(how="all") — только если ВСЕ NaN:')
print(df.dropna(how='all'), '\n')

print('dropna(subset=["A"]) — только по столбцу A:')
print(df.dropna(subset=['A']), '\n')

print('dropna(thresh=3) — оставить строки с ≥3 непустых:')
print(df.dropna(thresh=3))
````

````python
# === Заполнение пропусков ===
print('fillna(0):\n', df.fillna(0), '\n')

# Заполнение разными значениями для разных столбцов
print('fillna по столбцам:\n', df.fillna({'A': 0, 'B': df['B'].mean(), 'D': 'unknown'}), '\n')

# Заполнение предыдущим значением (forward fill)
print('ffill (вперёд):\n', df.ffill(), '\n')

# Заполнение следующим значением (backward fill)
print('bfill (назад):\n', df.bfill(), '\n')

# Интерполяция (для числовых)
print('interpolate (линейная):\n', df[['A', 'B', 'C']].interpolate())
````

<a id='12'></a>
## 12. Строковые операции (.str)

Аксессор `.str` даёт доступ к векторизованным строковым методам.

Документация: [Working with text data](https://pandas.pydata.org/docs/user_guide/text.html)

````python
s = pd.Series(['  Hello World  ', 'PYTHON pandas', 'Data Science 101', None, 'machine_learning'])

print('lower:\n', s.str.lower())
print('\nupper:\n', s.str.upper())
print('\ntitle:\n', s.str.title())
print('\nstrip:\n', s.str.strip())
print('\nlen:\n', s.str.len())
print('\ncontains "a" (регистронезависимо):\n', s.str.contains('a', case=False, na=False))
print('\nstartswith "D":\n', s.str.startswith('D', na=False))
````

````python
# === pipe — для цепочек операций ===
def add_ratio(df, col_a, col_b):
    return df.assign(ratio=df[col_a] / df[col_b])

def filter_big(df, threshold):
    return df[df['ratio'] > threshold]

result = (df
    .pipe(add_ratio, 'B', 'A')
    .pipe(filter_big, 8)
)
print('pipe chain:\n', result)
````

## 16. Группировка (groupby)

**Split → Apply → Combine** — самый важный паттерн в анализе данных.

Документация: [Group by](https://pandas.pydata.org/docs/user_guide/groupby.html)

````python
np.random.seed(42)
df = pd.DataFrame({
    'city': np.random.choice(['Москва', 'СПб', 'Казань'], 20),
    'department': np.random.choice(['IT', 'HR', 'Sales'], 20),
    'salary': np.random.randint(50, 150, 20) * 1000,
    'experience': np.random.randint(1, 15, 20),
})

print('Данные:', df.shape)
print(df.head(), '\n')

# Базовая группировка
print('=== Средняя зарплата по городам ===')
print(df.groupby('city')['salary'].mean(), '\n')

print('=== Несколько агрегаций ===')
print(df.groupby('city')['salary'].agg(['mean', 'median', 'min', 'max', 'count']), '\n')
````

````python
# Группировка по нескольким столбцам
print('По городу и отделу:')
print(df.groupby(['city', 'department'])['salary'].mean().round(0), '\n')
````

````python
# === agg — разные агрегации для разных столбцов ===
result = df.groupby('city').agg(
    avg_salary=('salary', 'mean'),
    max_salary=('salary', 'max'),
    avg_exp=('experience', 'mean'),
    count=('salary', 'count'),
).round(0)
print(result)
````

````python
# === transform — вернуть результат ТОГО ЖЕ РАЗМЕРА ===
# Полезно для нормализации внутри группы

df['salary_mean_city'] = df.groupby('city')['salary'].transform('mean')
df['salary_zscore'] = df.groupby('city')['salary'].transform(
    lambda x: (x - x.mean()) / x.std()
)
print(df[['city', 'salary', 'salary_mean_city', 'salary_zscore']].head(10))

# Убираем добавленные столбцы
df = df.drop(columns=['salary_mean_city', 'salary_zscore'])
````

````python
# === filter — оставить только группы, удовлетворяющие условию ===
# Оставить только города, где средняя зарплата > 90000
filtered = df.groupby('city').filter(lambda g: g['salary'].mean() > 90000)
print(f'До фильтра: {len(df)}, после: {len(filtered)}')
print(filtered.head())
````

````python
# === Итерация по группам ===
for name, group in df.groupby('city'):
    print(f'--- {name}: {len(group)} записей, средняя зп: {group["salary"].mean():.0f} ---')
````

## 18. Слияние таблиц (merge)

Аналог JOIN в SQL.

Документация: [Merge, join, concatenate](https://pandas.pydata.org/docs/user_guide/merging.html)

````python
# Две таблицы
orders = pd.DataFrame({
    'order_id': [1, 2, 3, 4, 5],
    'customer_id': [101, 102, 101, 103, 104],
    'amount': [250, 150, 300, 450, 200],
})

customers = pd.DataFrame({
    'customer_id': [101, 102, 103, 105],  # 104 нет, есть лишний 105
    'name': ['Алиса', 'Боб', 'Чарли', 'Фрэнк'],
    'city': ['Москва', 'СПб', 'Казань', 'Москва'],
})

print('Orders:', orders.shape)
print(orders, '\n')
print('Customers:', customers.shape)
print(customers)
````

````python
# === INNER JOIN — только совпадающие ===
inner = pd.merge(orders, customers, on='customer_id', how='inner')
print('INNER JOIN:')
print(inner, '\n')

# === LEFT JOIN — все из левой + совпадающие из правой ===
left = pd.merge(orders, customers, on='customer_id', how='left')
print('LEFT JOIN:')
print(left, '\n')

# === RIGHT JOIN ===
right = pd.merge(orders, customers, on='customer_id', how='right')
print('RIGHT JOIN:')
print(right, '\n')

# === OUTER JOIN — все из обеих ===
outer = pd.merge(orders, customers, on='customer_id', how='outer')
print('OUTER JOIN:')
print(outer)
````

````python
# Когда ключевые столбцы называются по-разному
df1 = pd.DataFrame({'id_заказа': [1, 2], 'сумма': [100, 200]})
df2 = pd.DataFrame({'order_id': [1, 2], 'product': ['A', 'B']})

merged = pd.merge(df1, df2, left_on='id_заказа', right_on='order_id')
print(merged)

# indicator — показать источник строки
m = pd.merge(orders, customers, on='customer_id', how='outer', indicator=True)
print('\nС индикатором:')
print(m[['customer_id', '_merge']])
````

````python
# === validate — проверка кардинальности ===
try:
    pd.merge(orders, customers, on='customer_id', validate='one_to_one')
except Exception as e:
    print(f'Ошибка: {e}')
    print('(customer_id=101 встречается дважды в orders)')
````

## 24. Лаговые признаки (shift, diff, pct_change)

Критически важны для прогнозирования временных рядов.

````python
data = pd.Series([100, 102, 105, 103, 108, 110], name='price',
                 index=pd.date_range('2024-01-01', periods=6, freq='D'))

df_lag = pd.DataFrame({
    'price': data,
    'lag_1': data.shift(1),             # значение вчера
    'lag_2': data.shift(2),             # позавчера
    'lead_1': data.shift(-1),           # значение завтра (в будущее)
    'diff_1': data.diff(1),             # price - lag_1 (изменение)
    'pct_change': data.pct_change(1),   # относительное изменение
})

print(df_lag)
````

````python
# Практика: создание лаговых признаков для ML
def create_lag_features(series, lags=[1, 2, 3, 7]):
    df = pd.DataFrame({'value': series})
    for lag in lags:
        df[f'lag_{lag}'] = series.shift(lag)
        df[f'diff_{lag}'] = series.diff(lag)
    df['rolling_mean_7'] = series.rolling(7, min_periods=1).mean()
    df['rolling_std_7'] = series.rolling(7, min_periods=1).std()
    return df

features = create_lag_features(ts)
print(features.head(10))
print(f'\nФорма: {features.shape}')
print(f'Пропуски:\n{features.isna().sum()}')
````

## 26. Categorical — экономия памяти

Для столбцов с малым числом уникальных значений (пол, город, статус).

## 27. Оптимизация памяти и быстрые форматы

Критически важно для больших датасетов.

````python
# === Функция автоматической оптимизации типов ===

def optimize_dtypes(df):
    """Уменьшает потребление памяти DataFrame."""
    start_mem = df.memory_usage(deep=True).sum() / 1024**2
    
    for col in df.columns:
        col_type = df[col].dtype
        
        if col_type == 'object':
            # Если мало уникальных значений — в category
            if df[col].nunique() / len(df) < 0.5:
                df[col] = df[col].astype('category')
        
        elif col_type in ['int64', 'int32']:
            c_min, c_max = df[col].min(), df[col].max()
            if c_min >= np.iinfo(np.int8).min and c_max <= np.iinfo(np.int8).max:
                df[col] = df[col].astype(np.int8)
            elif c_min >= np.iinfo(np.int16).min and c_max <= np.iinfo(np.int16).max:
                df[col] = df[col].astype(np.int16)
            elif c_min >= np.iinfo(np.int32).min and c_max <= np.iinfo(np.int32).max:
                df[col] = df[col].astype(np.int32)
        
        elif col_type == 'float64':
            df[col] = df[col].astype(np.float32)
    
    end_mem = df.memory_usage(deep=True).sum() / 1024**2
    print(f'Память: {start_mem:.2f} MB → {end_mem:.2f} MB ({100 * (1 - end_mem/start_mem):.0f}% экономия)')
    return df

# Тест
np.random.seed(42)
big_df = pd.DataFrame({
    'id': np.arange(100_000),
    'city': np.random.choice(['Москва', 'СПб', 'Казань'], 100_000),
    'value': np.random.randn(100_000),
    'count': np.random.randint(0, 100, 100_000),
})

big_df_opt = optimize_dtypes(big_df.copy())
print('\nТипы после оптимизации:')
print(big_df_opt.dtypes)
````

````python
# === Сравнение форматов хранения ===
import time
import os

test_df = pd.DataFrame({
    'a': np.random.randn(100_000),
    'b': np.random.randint(0, 100, 100_000),
    'c': np.random.choice(['x', 'y', 'z'], 100_000),
})

results = []

# CSV
start = time.perf_counter()
test_df.to_csv('test.csv', index=False)
t_write = time.perf_counter() - start
start = time.perf_counter()
pd.read_csv('test.csv')
t_read = time.perf_counter() - start
size = os.path.getsize('test.csv') / 1024
results.append(('CSV', t_write, t_read, size))

# Parquet
start = time.perf_counter()
test_df.to_parquet('test.parquet', index=False)
t_write = time.perf_counter() - start
start = time.perf_counter()
pd.read_parquet('test.parquet')
t_read = time.perf_counter() - start
size = os.path.getsize('test.parquet') / 1024
results.append(('Parquet', t_write, t_read, size))

# Feather
try:
    start = time.perf_counter()
    test_df.to_feather('test.feather')
    t_write = time.perf_counter() - start
    start = time.perf_counter()
    pd.read_feather('test.feather')
    t_read = time.perf_counter() - start
    size = os.path.getsize('test.feather') / 1024
    results.append(('Feather', t_write, t_read, size))
except Exception:
    pass

# Результаты
res_df = pd.DataFrame(results, columns=['Format', 'Write (s)', 'Read (s)', 'Size (KB)'])
print(res_df.to_string(index=False))

# Очистка
for f in ['test.csv', 'test.parquet', 'test.feather']:
    if os.path.exists(f):
        os.remove(f)
````

## 28. Полезные трюки и идиомы

````python
# === Method chaining — цепочки вызовов ===
np.random.seed(42)
raw = pd.DataFrame({
    'Name': ['  alice  ', 'BOB', 'Charlie', 'dave', None],
    'Score': [95, 87, None, 64, 91],
    'City': ['Moscow', 'SPB', 'Moscow', 'SPB', 'Kazan'],
})

clean = (raw
    .dropna(subset=['Name'])                          # убрать строки без имени
    .assign(Name=lambda x: x['Name'].str.strip().str.title())  # очистить имена
    .fillna({'Score': raw['Score'].median()})          # заполнить пропуски медианой
    .query('Score >= 70')                              # отфильтровать
    .sort_values('Score', ascending=False)             # отсортировать
    .reset_index(drop=True)                            # сбросить индекс
)

print('До:')
print(raw, '\n')
print('После:')
print(clean)
````
