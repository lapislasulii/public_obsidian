## 2. Два интерфейса: pyplot vs OOP

Matplotlib имеет **два способа** создания графиков:

1. **pyplot** (процедурный) — быстро, но ограниченно
2. **OOP (Figure/Axes)** — полный контроль, **рекомендуется**

Документация: [Usage Guide](https://matplotlib.org/stable/tutorials/introductory/quick_start.html)

````python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# === Способ 1: pyplot (MATLAB-стиль) ===
# Быстро для простых графиков, но путает при нескольких подграфиках
plt.figure(figsize=(8, 3))
plt.plot(x, y)
plt.title('pyplot стиль')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
# === Способ 2: OOP (рекомендуемый!) ===
# Явно создаём Figure и Axes — полный контроль

fig, ax = plt.subplots(figsize=(8, 3))   # fig — холст, ax — область рисования

ax.plot(x, y)                             # рисуем НА ax
ax.set_title('OOP стиль (рекомендуемый)')  # все методы через ax.set_*
ax.set_xlabel('x')
ax.set_ylabel('sin(x)')
ax.grid(True, alpha=0.3)

fig.tight_layout()
plt.show()
````

<a id='3'></a>
## 3. Архитектура Figure → Axes → Artists

```
Figure (фигура / холст)
├── Axes (область рисования) — может быть несколько
│   ├── XAxis (ось X)
│   │   ├── Label
│   │   ├── Ticks + TickLabels
│   ├── YAxis (ось Y)
│   ├── Title
│   ├── Lines, Patches, Text... (Artists)
│   └── Legend
├── Axes 2...
└── Suptitle (общий заголовок)
```

**Ключевое:** `Figure` ≠ `Axes`. На одном `Figure` может быть много `Axes`.

Документация: [Figure and Axes](https://matplotlib.org/stable/tutorials/introductory/quick_start.html#figure-and-axes)

````python
# Демонстрация иерархии
fig, ax = plt.subplots(figsize=(8, 4))

x = np.linspace(0, 10, 100)
line1, = ax.plot(x, np.sin(x), label='sin(x)')    # Line2D artist
line2, = ax.plot(x, np.cos(x), label='cos(x)')    # Line2D artist

ax.set_title('Заголовок Axes')      # Title artist
ax.set_xlabel('Ось X')              # XAxis Label
ax.set_ylabel('Ось Y')              # YAxis Label
ax.legend()                          # Legend artist
ax.grid(True, alpha=0.3)            # Grid lines

fig.suptitle('Заголовок Figure', fontsize=14, fontweight='bold', y=1.02)

# Инспекция
print(f'Figure: {type(fig)}')
print(f'Axes:   {type(ax)}')
print(f'Lines:  {ax.get_lines()}')
print(f'Title:  {ax.get_title()}')

fig.tight_layout()
plt.show()
````

<a id='4'></a>
## 4. Line plot (линейный график)

Основной тип для временных рядов, функций, трендов.

Документация: [ax.plot](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.plot.html)

````python
x = np.linspace(0, 4 * np.pi, 200)

fig, ax = plt.subplots(figsize=(10, 5))

# Несколько линий с разными стилями
ax.plot(x, np.sin(x), label='sin(x)', color='#2196F3', linewidth=2)
ax.plot(x, np.cos(x), label='cos(x)', color='#FF5722', linewidth=2, linestyle='--')
ax.plot(x, np.sin(x) * np.exp(-x/10), label='damped sin',
        color='#4CAF50', linewidth=2, linestyle='-.', alpha=0.8)

# Горизонтальная / вертикальная линия
ax.axhline(y=0, color='gray', linewidth=0.5)              # горизонтальная
ax.axvline(x=np.pi, color='gray', linewidth=0.5, ls=':')  # вертикальная

# Заливка между линиями
ax.fill_between(x, np.sin(x), np.cos(x), alpha=0.1, color='purple')

ax.set_title('Линейные графики', fontsize=14)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend(loc='upper right')
ax.grid(True, alpha=0.3)

fig.tight_layout()
plt.show()
````

<a id='5'></a>
## 5. Scatter plot (диаграмма рассеяния)

Для визуализации связи между двумя переменными.

Документация: [ax.scatter](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.scatter.html)

````python
np.random.seed(42)
n = 150
x = np.random.randn(n)
y = 0.5 * x + np.random.randn(n) * 0.5
colors = np.random.rand(n)       # цвет = числовая переменная
sizes = np.abs(np.random.randn(n)) * 200 + 20   # размер = ещё переменная

fig, ax = plt.subplots(figsize=(8, 6))

scatter = ax.scatter(x, y, c=colors, s=sizes,
                     cmap='viridis',         # палитра
                     alpha=0.6,
                     edgecolors='white',
                     linewidth=0.5)

# Colorbar
cbar = fig.colorbar(scatter, ax=ax)
cbar.set_label('Цветовая переменная')

ax.set_title('Scatter plot с 4 переменными (x, y, цвет, размер)')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.grid(True, alpha=0.3)

fig.tight_layout()
plt.show()
````

<a id='6'></a>
## 6. Bar plot (столбчатая диаграмма)

Документация: [ax.bar](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.bar.html)

````python
categories = ['Python', 'JavaScript', 'Java', 'C++', 'Go']
values_2023 = [35, 25, 20, 12, 8]
values_2024 = [38, 23, 18, 11, 10]

x_pos = np.arange(len(categories))
width = 0.35

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# --- Обычный ---
axes[0].bar(x_pos, values_2024, color='steelblue')
axes[0].set_xticks(x_pos)
axes[0].set_xticklabels(categories)
axes[0].set_title('Простой bar')
axes[0].set_ylabel('%')

# Подписи значений
for i, v in enumerate(values_2024):
    axes[0].text(i, v + 0.5, str(v), ha='center', fontweight='bold')

# --- Grouped ---
axes[1].bar(x_pos - width/2, values_2023, width, label='2023', color='#90CAF9')
axes[1].bar(x_pos + width/2, values_2024, width, label='2024', color='#1976D2')
axes[1].set_xticks(x_pos)
axes[1].set_xticklabels(categories)
axes[1].set_title('Grouped bar')
axes[1].legend()

# --- Horizontal ---
axes[2].barh(categories, values_2024, color='coral')
axes[2].set_title('Horizontal bar')
axes[2].set_xlabel('%')

fig.tight_layout()
plt.show()
````

````python
# === Stacked bar ===
fig, ax = plt.subplots(figsize=(8, 5))

ax.bar(categories, values_2023, label='2023', color='#BBDEFB')
ax.bar(categories, values_2024, bottom=values_2023, label='2024', color='#1976D2')

ax.set_title('Stacked bar')
ax.set_ylabel('Сумма %')
ax.legend()

fig.tight_layout()
plt.show()
````

<a id='7'></a>
## 7. Histogram (гистограмма)

Для визуализации распределения данных.

Документация: [ax.hist](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.hist.html)

````python
np.random.seed(42)
data_normal = np.random.randn(1000)
data_bimodal = np.concatenate([np.random.normal(-2, 0.8, 500),
                                np.random.normal(2, 1.0, 500)])

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Обычная гистограмма
axes[0].hist(data_normal, bins=30, color='steelblue', edgecolor='white', alpha=0.8)
axes[0].set_title('Гистограмма (bins=30)')
axes[0].axvline(np.mean(data_normal), color='red', ls='--', label=f'mean={np.mean(data_normal):.2f}')
axes[0].legend()

# Нормализованная (density)
axes[1].hist(data_normal, bins=30, density=True, color='lightcoral', edgecolor='white')
# Наложение KDE
x_kde = np.linspace(-4, 4, 100)
from scipy.stats import norm
try:
    axes[1].plot(x_kde, norm.pdf(x_kde), 'k-', lw=2, label='N(0,1)')
    axes[1].legend()
except ImportError:
    pass
axes[1].set_title('density=True + PDF')

# Два распределения
axes[2].hist(data_normal, bins=30, alpha=0.5, label='Normal', color='blue')
axes[2].hist(data_bimodal, bins=30, alpha=0.5, label='Bimodal', color='orange')
axes[2].set_title('Два распределения')
axes[2].legend()

fig.tight_layout()
plt.show()
````

## 8. Pie chart и другие типы

````python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# --- Pie chart ---
sizes = [35, 25, 20, 12, 8]
labels = ['Python', 'JS', 'Java', 'C++', 'Go']
explode = (0.05, 0, 0, 0, 0)  # выдвинуть первый сектор

axes[0].pie(sizes, labels=labels, explode=explode, autopct='%1.1f%%',
            colors=sns.color_palette('Set2'), startangle=90)
axes[0].set_title('Pie chart')

# --- Errorbar ---
x = np.arange(5)
y = [20, 35, 30, 25, 40]
yerr = [2, 3, 4, 2, 3]
axes[1].errorbar(x, y, yerr=yerr, fmt='o-', capsize=5, color='steelblue')
axes[1].set_title('Error bars')
axes[1].set_xticks(x)
axes[1].set_xticklabels(['A', 'B', 'C', 'D', 'E'])

# --- Step plot ---
x = np.arange(10)
y = np.random.randint(1, 10, 10)
axes[2].step(x, y, where='mid', color='green', linewidth=2)
axes[2].fill_between(x, y, step='mid', alpha=0.2, color='green')
axes[2].set_title('Step plot')

fig.tight_layout()
plt.show()
````

## 9. Subplots — несколько графиков

Документация: [subplots](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html)

````python
# === Простая сетка ===
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

# axes — 2D массив (2×3)
x = np.linspace(0, 2*np.pi, 100)
funcs = [np.sin, np.cos, np.tan, lambda x: np.sin(2*x), lambda x: np.cos(2*x), np.abs]
names = ['sin', 'cos', 'tan', 'sin(2x)', 'cos(2x)', '|x|']

for ax, func, name in zip(axes.flatten(), funcs, names):
    y = func(x)
    y = np.clip(y, -5, 5)  # ограничить tan
    ax.plot(x, y, linewidth=2)
    ax.set_title(name, fontsize=12)
    ax.grid(True, alpha=0.3)

fig.suptitle('Сетка 2×3 subplots', fontsize=16, fontweight='bold')
fig.tight_layout()
plt.show()
````

````python
# === Общие оси ===
fig, axes = plt.subplots(2, 2, figsize=(10, 8),
                         sharex=True,    # общая ось X
                         sharey='row')   # общая Y по строкам

np.random.seed(42)
for ax in axes.flatten():
    ax.plot(np.cumsum(np.random.randn(100)))

fig.suptitle('Shared axes', fontsize=14)
fig.tight_layout()
plt.show()
````

````python
# === Разные размеры: gridspec ===
fig = plt.figure(figsize=(12, 6))

# GridSpec: 2 строки, 3 столбца
gs = fig.add_gridspec(2, 3, hspace=0.3, wspace=0.3)

ax_big = fig.add_subplot(gs[0, :])       # верхняя строка, все столбцы
ax_bl  = fig.add_subplot(gs[1, 0])       # нижний левый
ax_bm  = fig.add_subplot(gs[1, 1])       # нижний средний
ax_br  = fig.add_subplot(gs[1, 2])       # нижний правый

x = np.linspace(0, 10, 200)
ax_big.plot(x, np.sin(x) * np.exp(-x/5), linewidth=2, color='steelblue')
ax_big.set_title('Большой график (весь верх)', fontsize=12)

ax_bl.hist(np.random.randn(500), bins=20, color='coral')
ax_bl.set_title('Hist')

ax_bm.scatter(np.random.rand(50), np.random.rand(50), alpha=0.6)
ax_bm.set_title('Scatter')

ax_br.bar(['A', 'B', 'C'], [3, 7, 5], color='#4CAF50')
ax_br.set_title('Bar')

plt.show()
````

````python
# === subplot_mosaic — самый удобный способ (matplotlib >= 3.3) ===
fig, axes = plt.subplot_mosaic(
    [['top', 'top', 'top'],
     ['bl',  'bm',  'br']],
    figsize=(12, 6)
)

axes['top'].plot(np.sin(np.linspace(0, 10, 200)), color='steelblue', lw=2)
axes['top'].set_title('Top')
axes['bl'].hist(np.random.randn(300), bins=20, color='coral')
axes['bm'].scatter(np.random.rand(50), np.random.rand(50))
axes['br'].bar(['X', 'Y', 'Z'], [5, 3, 8], color='green')

fig.tight_layout()
plt.show()
````

## 10. Оформление: заголовки, подписи, легенда, сетка

````python
fig, ax = plt.subplots(figsize=(10, 6))

x = np.linspace(0, 10, 100)
ax.plot(x, np.sin(x), label='sin(x)', linewidth=2)
ax.plot(x, np.cos(x), label='cos(x)', linewidth=2)

# Заголовки
ax.set_title('Тригонометрические функции', fontsize=16, fontweight='bold', pad=15)
ax.set_xlabel('Время (сек)', fontsize=12, labelpad=10)
ax.set_ylabel('Амплитуда', fontsize=12, labelpad=10)

# Легенда — много настроек
ax.legend(
    loc='upper right',        # позиция: 'best', 'upper left', 'center', ...
    fontsize=11,
    framealpha=0.9,            # прозрачность фона
    edgecolor='gray',          # цвет рамки
    ncol=2,                    # число колонок
    title='Функции',           # заголовок легенды
    title_fontsize=12,
)

# Сетка
ax.grid(True, which='major', linestyle='-', alpha=0.3)
ax.grid(True, which='minor', linestyle=':', alpha=0.15)
ax.minorticks_on()  # включить мелкие деления

fig.tight_layout()
plt.show()
````

## 11. Стили, цвета, маркеры, линии

Документация: [Styling](https://matplotlib.org/stable/tutorials/introductory/customizing.html)

````python
# === Встроенные стили ===
print('Доступные стили:', plt.style.available)
````

````python
# Использование стиля
x = np.linspace(0, 10, 100)

styles = ['default', 'seaborn-v0_8-whitegrid', 'ggplot', 'dark_background']

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, style_name in zip(axes.flatten(), styles):
    with plt.style.context(style_name):    # временно применить стиль
        ax.plot(x, np.sin(x), linewidth=2)
        ax.plot(x, np.cos(x), linewidth=2)
        ax.set_title(f'Style: {style_name}', fontsize=10)
        ax.grid(True, alpha=0.3)

fig.tight_layout()
plt.show()
````

````python
# === Цвета ===
fig, ax = plt.subplots(figsize=(10, 6))
x = np.linspace(0, 5, 50)

# Разные способы задать цвет
ax.plot(x, x,     label='named: red',      color='red')
ax.plot(x, x+1,   label='hex: #FF8C00',    color='#FF8C00')
ax.plot(x, x+2,   label='RGB tuple',       color=(0.2, 0.6, 0.3))
ax.plot(x, x+3,   label='CSS: steelblue',  color='steelblue')
ax.plot(x, x+4,   label='C0 (цикл)',       color='C0')    # цвет из текущей палитры
ax.plot(x, x+5,   label='tab:purple',      color='tab:purple')

ax.set_title('Способы задания цвета')
ax.legend()
fig.tight_layout()
plt.show()
````

````python
# === Маркеры и стили линий ===
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
x = np.arange(0, 10, 0.5)

# Маркеры
markers = ['o', 's', '^', 'D', 'v', 'p', '*', 'X']
for i, m in enumerate(markers):
    axes[0].plot(x, np.sin(x + i*0.3) + i*0.5, marker=m, markersize=6, label=f"'{m}'")
axes[0].set_title('Маркеры')
axes[0].legend(ncol=2, fontsize=8)

# Стили линий
linestyles = ['-', '--', '-.', ':', (0, (5, 10)), (0, (3, 1, 1, 1))]
names = ['solid', 'dashed', 'dashdot', 'dotted', 'custom1', 'custom2']
for i, (ls, name) in enumerate(zip(linestyles, names)):
    axes[1].plot(x, np.ones_like(x) * (len(linestyles) - i),
                linestyle=ls, linewidth=2, label=name, color=f'C{i}')
axes[1].set_title('Стили линий')
axes[1].legend()
axes[1].set_ylim(0, 7)

fig.tight_layout()
plt.show()
````

````python
# === Colormaps (палитры) ===
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

np.random.seed(42)
data = np.random.randn(10, 10)
cmaps = ['viridis', 'plasma', 'coolwarm', 'RdYlGn', 'Blues', 'Set2']

for ax, cmap in zip(axes.flatten(), cmaps):
    im = ax.imshow(data, cmap=cmap, aspect='auto')
    ax.set_title(cmap)
    fig.colorbar(im, ax=ax, shrink=0.8)

fig.suptitle('Colormaps', fontsize=14)
fig.tight_layout()
plt.show()
````

## 12. Аннотации и текст

Документация: [Annotations](https://matplotlib.org/stable/tutorials/text/annotations.html)

````python
fig, ax = plt.subplots(figsize=(10, 6))

x = np.linspace(0, 2*np.pi, 200)
y = np.sin(x)
ax.plot(x, y, linewidth=2, color='steelblue')

# Простой текст
ax.text(4.5, 0.5, 'Простой текст', fontsize=12, color='gray')

# Текст с рамкой
ax.text(0.5, -0.5, 'С рамкой', fontsize=12,
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

# Аннотация со стрелкой
ax.annotate('Максимум',
            xy=(np.pi/2, 1),              # куда указывает стрелка
            xytext=(np.pi/2 + 1, 0.7),    # где текст
            fontsize=12,
            arrowprops=dict(
                arrowstyle='->',
                color='red',
                connectionstyle='arc3,rad=0.3'
            ),
            color='red')

# Аннотация минимума
ax.annotate('Минимум',
            xy=(3*np.pi/2, -1),
            xytext=(3*np.pi/2 + 1, -0.5),
            fontsize=12,
            arrowprops=dict(arrowstyle='->', color='blue'),
            color='blue')

# Затенённая область
ax.axvspan(np.pi, 2*np.pi, alpha=0.1, color='red', label='Отрицательная область')

ax.set_title('Аннотации и текст')
ax.legend()
ax.grid(True, alpha=0.3)

fig.tight_layout()
plt.show()
````

<a id='13'></a>
## 13. Оси: пределы, шкалы, тики

Документация: [Axis ticks](https://matplotlib.org/stable/gallery/ticks/index.html)

## 21. Фасетные сетки: FacetGrid, catplot, relplot, displot

**Figure-level функции** Seaborn создают собственный `Figure` с несколькими `Axes`, разбитыми по категориям.

Документация: [Figure-level vs axes-level](https://seaborn.pydata.org/tutorial/function_overview.html)

````python
# === Figure-level vs Axes-level ===
# Figure-level: relplot, displot, catplot — создают свой FacetGrid
# Axes-level:   scatterplot, histplot, boxplot... — рисуют на данном ax

# relplot — обёртка над scatterplot/lineplot с фасетами
g = sns.relplot(
    data=tips,
    x='total_bill', y='tip',
    hue='smoker',           # цвет
    col='time',             # столбцы фасетов
    row='sex',              # строки фасетов
    kind='scatter',         # 'scatter' или 'line'
    height=4, aspect=1.2,
)
g.fig.suptitle('relplot с фасетами', y=1.02, fontsize=14)
plt.show()
````

````python
# === displot — figure-level для распределений ===
g = sns.displot(
    data=tips,
    x='total_bill',
    hue='sex',
    col='time',
    kind='kde',            # 'hist', 'kde', 'ecdf'
    fill=True,
    height=4, aspect=1.3,
)
g.fig.suptitle('displot (KDE по времени)', y=1.02)
plt.show()
````

````python
# === catplot — figure-level для категорий ===
g = sns.catplot(
    data=tips,
    x='day', y='total_bill',
    hue='sex',
    col='time',
    kind='violin',          # 'strip', 'swarm', 'box', 'violin', 'bar', 'count', 'point'
    split=True,
    height=5, aspect=1.0,
)
g.fig.suptitle('catplot (violin, split по полу)', y=1.02)
plt.show()
````

````python
# === FacetGrid напрямую — максимальная гибкость ===
g = sns.FacetGrid(tips, col='day', col_wrap=2, height=4, aspect=1.2)
g.map_dataframe(sns.scatterplot, x='total_bill', y='tip', hue='sex', alpha=0.6)
g.add_legend()
g.set_titles('День: {col_name}')
g.fig.suptitle('FacetGrid (ручной)', y=1.02, fontsize=14)
plt.show()
````

````python
# === col_wrap — перенос на новую строку ===
g = sns.catplot(
    data=tips,
    x='sex', y='total_bill',
    col='day',
    col_wrap=2,              # 2 столбца → остальные переносятся
    kind='box',
    height=4, aspect=1.2,
)
g.fig.suptitle('col_wrap=2', y=1.02)
plt.show()
````

## 24. Кастомизация Seaborn-графиков через Matplotlib

Seaborn возвращает объекты Matplotlib, поэтому кастомизировать можно всё.

````python
# === Axes-level: передаём ax ===
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Seaborn рисует на указанном ax
sns.histplot(data=tips, x='total_bill', kde=True, ax=axes[0])
axes[0].set_title('Мой заголовок', fontsize=14, color='navy')
axes[0].set_xlabel('Счёт ($)', fontsize=12)
axes[0].axvline(tips['total_bill'].mean(), color='red', ls='--', label='Среднее')
axes[0].legend()

sns.boxplot(data=tips, x='day', y='total_bill', ax=axes[1])
axes[1].set_title('Счёт по дням', fontsize=14)
axes[1].set_ylabel('Счёт ($)')

fig.suptitle('Seaborn + Matplotlib кастомизация', fontsize=16, fontweight='bold')
fig.tight_layout()
plt.show()
````

````python
# === Рецепт 3: Временной ряд с выделением аномалий ===
np.random.seed(42)
dates = pd.date_range('2024-01-01', periods=100, freq='D')
values = np.cumsum(np.random.randn(100)) + 50
# Добавим "аномалии"
values[30] = 80
values[65] = 20

ts = pd.DataFrame({'date': dates, 'value': values})
ts['rolling_mean'] = ts['value'].rolling(7).mean()
ts['rolling_std'] = ts['value'].rolling(7).std()
ts['upper'] = ts['rolling_mean'] + 2 * ts['rolling_std']
ts['lower'] = ts['rolling_mean'] - 2 * ts['rolling_std']
ts['anomaly'] = (ts['value'] > ts['upper']) | (ts['value'] < ts['lower'])

fig, ax = plt.subplots(figsize=(14, 5))

ax.plot(ts['date'], ts['value'], label='Значение', alpha=0.7)
ax.plot(ts['date'], ts['rolling_mean'], label='MA(7)', color='orange', lw=2)
ax.fill_between(ts['date'], ts['lower'], ts['upper'], alpha=0.15, color='orange', label='±2σ')

# Выделяем аномалии
anomalies = ts[ts['anomaly']]
ax.scatter(anomalies['date'], anomalies['value'],
           color='red', s=100, zorder=5, label='Аномалии', edgecolors='darkred')

ax.set_title('Детекция аномалий (±2σ)', fontsize=14)
ax.legend()
ax.grid(True, alpha=0.3)
fig.autofmt_xdate()
fig.tight_layout()
plt.show()
````

````python
# === Рецепт 4: Матрица пропусков ===
np.random.seed(42)
df_miss = pd.DataFrame(
    np.random.choice([1, np.nan], size=(30, 8), p=[0.7, 0.3]),
    columns=[f'col_{i}' for i in range(8)]
)

fig, ax = plt.subplots(figsize=(10, 6))
sns.heatmap(df_miss.isna(), cbar=False, cmap='YlOrRd',
            yticklabels=False, ax=ax)
ax.set_title('Матрица пропусков (жёлтый = NaN)', fontsize=14)
fig.tight_layout()
plt.show()
````

````python
# === Рецепт 5: Confusion matrix heatmap ===
from sklearn.metrics import confusion_matrix
try:
    y_true = np.random.choice(['cat', 'dog', 'bird'], 200, p=[0.5, 0.3, 0.2])
    y_pred = y_true.copy()
    # Добавим ошибки
    mask = np.random.rand(200) < 0.2
    y_pred[mask] = np.random.choice(['cat', 'dog', 'bird'], mask.sum())

    labels = ['bird', 'cat', 'dog']
    cm = confusion_matrix(y_true, y_pred, labels=labels)

    fig, ax = plt.subplots(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=labels, yticklabels=labels, ax=ax)
    ax.set_xlabel('Predicted')
    ax.set_ylabel('Actual')
    ax.set_title('Confusion Matrix', fontsize=14)
    fig.tight_layout()
    plt.show()
except ImportError:
    print('sklearn не установлен — пропускаем этот пример')
````

````python
# === Рецепт 6: Feature importance ===
features = ['score', 'age', 'experience', 'education', 'salary',
             'distance', 'rating', 'reviews']
importances = np.array([0.25, 0.18, 0.15, 0.12, 0.10, 0.08, 0.07, 0.05])

# Сортировка
idx = np.argsort(importances)

fig, ax = plt.subplots(figsize=(8, 5))
ax.barh(np.array(features)[idx], importances[idx], color='steelblue')
ax.set_xlabel('Importance')
ax.set_title('Feature Importance', fontsize=14)

# Подписи
for i, v in enumerate(importances[idx]):
    ax.text(v + 0.005, i, f'{v:.2f}', va='center')

fig.tight_layout()
plt.show()
````
